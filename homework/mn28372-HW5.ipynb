```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0,1.01,0.01)

m_real = np.random.uniform(1,5)
b_real = np.random.uniform(0,3)

error = np.random.uniform(-1,1,len(x))

y = m_real*x+b_real+error
x = np.sort(x)
n = len(x)

sumx = np.sum(x)
sumy = np.sum(y)
sumx2 = np.sum(x*x)
sumxy = np.sum(x*y)

m = (n*sumxy-sumx*sumy)/(n*sumx2-sumx**2)
b = (sumy-m*sumx)/n

A = np.vstack([x, np.ones(len(x))]).T
m_np = np.linalg.lstsq(A,y,rcond=None)[0]
b_np = np.linalg.lstsq(A,y,rcond=None)[0]

print("Original Slope:",m_real)
print("Original b:",b_real)

print("Own Least Squares Slope:",m)
print("Own Least Squares b:",b)

print("Numpy Slope:",m_np)
print("Numpy b:",b_np)

yp = m*x + b

residuals = y-yp

print("Residuals:")
print(residuals)

max_res = np.max(residuals)
min_res = np.min(residuals)

max_index = np.argmax(residuals)
min_index = np.argmin(residuals)

print("Biggest residual:",max_res,"where x=",x[max_index])
print("Smallest residual:",min_res,"where x=",x[min_index])

plt.scatter(x,y)

xline = np.linspace(min(x),max(x),100)
yline = m*xline + b

plt.plot(xline,yline)

plt.show()
```
