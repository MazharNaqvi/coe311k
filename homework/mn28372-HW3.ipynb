```python
import numpy as np
import matplotlib.pyplot as plt
import math

def taylor(x,nterms):
    result = 0
    for n in range(nterms):
        term = ((-1)**n)*(x**(2*n+1))/math.factorial(2*n+1)
        result = result+term
    return result
xvalues=np.arange(0,10.1,0.1)

e1 = []
e2 = []
e3 = []
e4 = []

for x in xvalues:
    exactsin = np.sin(x)
    estimate1 = taylor(x,1)
    estimate2 = taylor(x,2)
    estimate3 = taylor(x,3)
    estimate4 = taylor(x,4)
    
    e1.append(abs(exactsin-estimate1))
    e2.append(abs(exactsin-estimate2))
    e3.append(abs(exactsin-estimate3))
    e4.append(abs(exactsin-estimate4))

plt.plot(xvalues,e1,label="1")
plt.plot(xvalues,e2,label="2")
plt.plot(xvalues,e3,label="3")
plt.plot(xvalues,e4,label="4")
plt.title("Truncation Errors for Taylor Series of sin(x) Function")
plt.xlabel("X")
plt.ylabel("Error")
plt.legend()
plt.show
```

```python
#The truncation error does not decrease uniformly of rall values of x
#The truncation error graph is steeper for larger values of x
```
