```python
import random
import math
import matplotlib.pyplot as plt

def approximatePi(numSamples):
    counter = 0
    
    for s in range(numSamples):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        
        if x * x + y * y < 1:
            counter = counter + 1
    
    return 4.0 * counter / numSamples

r_pi = round(math.pi, 15)

sample_size = []
errors = []

for n in range(10, 10000, 20):
    pi_estimate = approximatePi(n)
    error = abs(pi_estimate - r_pi)
    
    sample_size.append(n)
    errors.append(error)
    
plt.figure()
plt.plot(sample_size, errors)
plt.xlabel("Number of Samples")
plt.ylabel("Errors")
plt.title("Monte Carlo Appoximation of Pi")
plt.grid(True)
plt.show()
```
