```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return -1000*y+3000-2000*np.exp(-x)

def exact(x):
    return 3-0.998*np.exp(-1000*x)-2.002*np.exp(-x)

h = 0.001

x0 = 0
x_end = 0.02

steps = int((x_end - x0)/h)

x = [0]*(steps+1)
y = [0]*(steps+1)

x[0] = x0
y[0] = 0

for i in range(steps):
    x[i+1] = x[i]+h
    y[i+1] = y[i]+h*f(x[i], y[i])

x = np.array(x)
y = np.array(y)

y_exact = exact(x)

plt.plot(x, y, label="Euler")
plt.plot(x, y_exact, label="Exact")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Euler vs Exact Solution")
plt.legend()
plt.grid()
plt.show()
```
