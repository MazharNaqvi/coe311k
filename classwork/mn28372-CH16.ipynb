```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return -2*y

def exact(x):
    return np.exp(-2*x)

def euler(h):
    x_start = 0
    x_end = 2
    num_steps = int((x_end-x_start)/h)

    x_values = np.linspace(x_start, x_end, num_steps+1)
    y_values = np.zeros(num_steps+1)

    y_values[0] = 1

    for i in range(num_steps):
        y_values[i+1] = y_values[i]+h*f(x_values[i], y_values[i])

    return x_values, y_values

h_values = [0.1, 0.5, 1.0, 1.5]

for h in h_values:
    x_values, y_values = euler(h)
    exact_values = exact(x_values)

    local_error = np.abs(exact_values - y_values)
    global_error = np.max(local_error)

    print("h =", h)
    print("local error =", local_error)
    print("global error =", global_error)

    plt.plot(x_values, local_error, marker='o', label="h = " + str(h))

plt.xlabel("x")
plt.ylabel("error")
plt.title("Error for Euler's Method")
plt.legend()
plt.show()
```
