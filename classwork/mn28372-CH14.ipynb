```python
import numpy as np
import matplotlib.pyplot as plt

a = 0
b = 2 * np.pi
N = 20

x = np.linspace(a, b, N)
f = np.sin

h = (b - a) / (N - 1)

df_central = (f(x + h) - f(x - h)) / (2 * h)
df_forward = (f(x + h) - f(x)) / h
df_backward = (f(x) - f(x - h)) / h
true_derivative = np.cos(x)

plt.plot(x, true_derivative, label="cos(x)")
plt.plot(x, df_forward, label="Forward Difference")
plt.plot(x, df_backward, label="Backward Difference")
plt.plot(x, df_central, label="Central Difference")

plt.title("Derivative Approximations of sin(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```
