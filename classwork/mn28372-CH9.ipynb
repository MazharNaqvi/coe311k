```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

A = np.vstack([X, np.ones(len(X))]).T

m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print("Slope:", m)
print("Intercept:", c)

xline = np.linspace(min(X), max(X), 100)
yline = m * xline + c

plt.scatter(X, Y)
plt.plot(xline, yline)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

np.random.shuffle(X)
np.random.shuffle(Y)

A = np.vstack([X, np.ones(len(X))]).T

m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print("Slope:", m)
print("Intercept:", c)

xline = np.linspace(min(X), max(X), 100)
yline = m * xline + c

plt.scatter(X, Y)
plt.plot(xline, yline)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 1.01, 0.01)

real_m = 3
real_c = 2

noise = np.random.normal(0, 0.2, len(x))

y = real_m * x + real_c + noise

A = np.vstack([x, np.ones(len(x))]).T

m, c = np.linalg.lstsq(A, y, rcond=None)[0]

print("Original slope:", real_m)
print("Original intercept:", real_c)
print("Fitted slope:", m)
print("Fitted intercept:", c)

xline = np.linspace(min(x), max(x), 100)
yline = m * xline + c

plt.scatter(x, y)
plt.plot(xline, yline)
plt.show()
```
