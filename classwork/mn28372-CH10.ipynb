```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

x = np.arange(0,1.01,0.01)

m = 3
b = 2

error = np.random.uniform(-1,1,len(x))

y = m*x+b+error


A = np.vstack([x, np.ones(len(x))]).T
m_fit, b_fit = np.linalg.lstsq(A,y,rcond=None)[0]

y_line = m_fit*x+b_fit


degree = 3
coeff = np.polyfit(x,y,degree)

poly = np.poly1d(coeff)

y_poly = poly(x)


def mse(y_true,y_pred):
    return np.mean((y_true-y_pred)**2)

def r2(y_true,y_pred):
    ss_res = np.sum((y_true-y_pred)**2)
    ss_tot = np.sum((y_true-np.mean(y_true))**2)
    return 1-ss_res/ss_tot


mse_line = mse(y,y_line)
r2_line = r2(y,y_line)

mse_poly = mse(y,y_poly)
r2_poly = r2(y,y_poly)

print("Linear Fit MSE:",mse_line)
print("Linear Fit R2:",r2_line)

print("Polynomial Fit MSE:",mse_poly)
print("Polynomial Fit R2:",r2_poly)

if mse_line < mse_poly:
    print("Linear fit better")
else:
    print("Polynomial fit better")


plt.scatter(x,y,label="Data Points")
plt.plot(x,y_line,label="Least Squares Line")
plt.plot(x,y_poly,label="Cubic Polynomial")

plt.title("Linear vs Polynomial Fit")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()



print("\nTest diff polynomial degrees")

for d in range(1,10):

    coeff = np.polyfit(x,y,d)

    poly = np.poly1d(coeff)

    y_pred = poly(x)

    m = mse(y,y_pred)
    r = r2(y,y_pred)

    print("Degree:",d," MSE:",m," R2:",r)



degree = 3

V = np.vander(x,degree+1,increasing=True)

coeff_v = np.linalg.lstsq(V,y,rcond=None)[0]

y_v = V @ coeff_v

mse_v = mse(y,y_v)
r2_v = r2(y,y_v)

print("\nVandermonde Results")
print("MSE:",mse_v)
print("R2:",r2_v)

plt.scatter(x,y,label="Data Points")
plt.plot(x,y_v,label="Vandermonde Fit")

plt.title("Vandermonde Polynomial Fit")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
