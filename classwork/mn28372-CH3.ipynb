```python
import random
import math
import matplotlib.pyplot as plt

def approximatePi(numSamples):
    counter = 0
    
    for s in range(numSamples):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        
        if x * x + y * y < 1:
            counter = counter + 1
    
    return 4.0 * counter / numSamples

runs = 100
samples = 1000

absolute_errors = []
relative_errors = []

rpi= math.pi

for s in range(runs):
    pi_estimate = approximatePi(samples)
    
    abs_error = abs(rpi - pi_estimate)
    rel_error = abs_error / rpi
    
    absolute_errors.append(abs_error)
    relative_errors.append(rel_error)

plt.plot(absolute_errors)
plt.xlabel("Runs")
plt.ylabel("Absolute Error")
plt.title("Absolute Error for Pi Estimate")
plt.show()

plt.plot(relative_errors)
plt.xlabel("Runs")
plt.ylabel("Relative Error")
plt.title("Relative Error for Pi Estimate")
plt.show()
```
