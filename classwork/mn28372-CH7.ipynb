```python
import numpy as np
import time

def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    
    start_time = time.time()
    
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j]*x[j] for j in range(n) if j != i)
            x_new[i] = (b[i]-s)/A[i][i]
            
        if np.linalg.norm(x_new-x, ord=np.inf) < tol:
            end_time = time.time()
            return x_new, k+1, end_time-start_time
        x = x_new
    
    end_time = time.time()
    return x, max_iterations, end_time - start_time


def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    start_time = time.time()
    
    for k in range(max_iterations):
        x_old = x.copy()
        
        for i in range(n):
            s1 = sum(A[i][j]*x[j] for j in range(i))
            s2 = sum(A[i][j]*x_old[j] for j in range(i+1, n))
            x[i] = (b[i]-s1-s2)/A[i][i]
        
        if np.linalg.norm(x-x_old, ord=np.inf) < tol:
            end_time = time.time()
            return x, k+1, end_time-start_time
    
    end_time = time.time()
    return x, max_iterations, end_time - start_time


tol = 1e-6
max_iterations = 10000

# Part 1
print("\nPart 1")

A = np.array([
[0.582745, 0.48 , 0.10 , 0.  , 0.  ],
[0.48 , 1.044129, 0.46 , 0.10 , 0.  ],
[0.10 , 0.46 , 1.10431 , 0.44 , 0.10 ],
[0.  , 0.10 , 0.44 , 0.963889, 0.42 ],
[0.  , 0.  , 0.10 , 0.42 , 0.522565]
], dtype=float)

b = np.array([1.162745, 2.084129, 2.20431 , 1.923889, 1.042565], dtype=float)

x0 = np.zeros_like(b)

j_solution, j_iterations, j_time = jacobi(A, b, x0, tol, max_iterations)
print("Jacobi Solution:", j_solution)
print("Jacobi Iterations:", j_iterations)
print("Jacobi Time:", j_time)

g_solution, g_iterations, g_time = gauss_seidel(A, b, x0, tol, max_iterations)
print("Gauss-Seidel Solution:", g_solution)
print("Gauss-Seidel Iterations:", g_iterations)
print("Gauss-Seidel Time:", g_time)

# Part 2
print("\nPart 2")

A = np.array([[1, 1], [1, 1.0001]], dtype=float)
b = np.array([1, 1], dtype=float)

x0 = np.zeros_like(b)

j_solution, j_iterations, j_time = jacobi(A, b, x0, tol, max_iterations)
print("Jacobi Solution:", j_solution)
print("Jacobi Iterations:", j_iterations)
print("Jacobi Time:", j_time)

g_solution, g_iterations, g_time = gauss_seidel(A, b, x0, tol, max_iterations)
print("Gauss-Seidel Solution:", g_solution)
print("Gauss-Seidel Iterations:", g_iterations)
print("Gauss-Seidel Time:", g_time)

# Part 3
print("\nPart 3")

A = np.array([[1, 1], [1, 1.0001]], dtype=float)
b = np.array([1, 1.0001], dtype=float)

x0 = np.zeros_like(b)

j_solution, j_iterations, j_time = jacobi(A, b, x0, tol, max_iterations)
print("Solution:", j_solution)
print("Iterations:", j_iterations)
print("Time:", j_time)

g_solution, g_iterations, g_time = gauss_seidel(A, b, x0, tol, max_iterations)
print("Solution:", g_solution)
print("Iterations:", g_iterations)
print("Time:", g_time)
```
