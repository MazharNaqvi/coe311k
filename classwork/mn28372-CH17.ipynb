```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([[-2, 0],
              [0, -3]])

y_start = np.array([1, 1])

t_start = 0
t_end = 5

h_values = [0.1, 0.5, 0.9, 1.1]

for h in h_values:

    steps = int((t_end-t_start)/h)

    t = np.zeros(steps+1)
    y = np.zeros((steps+1, 2))

    t[0] = t_start
    y[0] = y_start

    I = np.eye(2)
    E = I+h*A

    eig = np.linalg.eigvals(E)

    print("h =", h)

    if np.all(np.abs(eig) < 1):
        print("Solution is Stable")
    else:
        print("Solution may not be Entirely Stable ;)")

    print()

    for i in range(steps):
        t[i+1] = t[i]+h
        y[i+1] = np.dot(E, y[i])

    plt.plot(t, y[:, 0], label="y1 (h="+str(h)+")")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Euler Method Solutions")
plt.legend()
plt.grid()
plt.show()
```
