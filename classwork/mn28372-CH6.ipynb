```python
import numpy as np
import time

def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    
    start_time = time.time()
    
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j]*x[j] for j in range(n) if j != i)
            x_new[i] = (b[i]-s)/A[i][i]
            
        if np.linalg.norm(x_new-x, ord=np.inf) < tol:
            end_time = time.time()
            return x_new, k+1, end_time-start_time
        x = x_new
    
    end_time = time.time()
    return x, max_iterations, end_time - start_time

def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    start_time = time.time()
    
    for k in range(max_iterations):
        x_new = x.copy()
        
        for i in range(n):
            s1 = sum(A[i][j]*x_new[j] for j in range(i))
            s2 = sum(A[i][j]*x[j] for j in range(i+1, n))
            x_new[i] = (b[i]-s1-s2)/A[i][i]
            
            if np.linalg.norm(x_new-x, ord=np.inf) < tol:
                end_time = time.time()
                return x_new, k+1, end_time-start_time
            x = x_new
            
        end_time = time.time()
        return x, max_iterations, end_time - start_time

A = np.array([
    [10, -1,  2,  0,  0],
    [-1, 11, -1,  3,  0],
    [ 2, -1, 10, -1,  0],
    [ 0,  3, -1,  8, -2],
    [ 0,  0,  0, -2,  9]
])

b = np.array([14, 30, 26, 25, 37])

x0 = np.zeros_like(b)
tol = 1e-6
max_iterations = 10000

j_solution, j_iterations, j_time = jacobi(A, b, x0, tol, max_iterations)
print(f"Solution: {j_solution}")
print(f"Iterations: {j_iterations}")
print(f"Time: {j_time}")

g_solution, g_iterations, g_time = gauss_seidel(A, b, x0, tol, max_iterations)
print(f"Solution: {g_solution}")
print(f"Iterations: {g_iterations}")
print(f"Time: {g_time}")
```
