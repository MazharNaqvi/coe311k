```python
import numpy as np

pi_estimate = 0
terms = 5000

for n in range(terms):
    fraction = ((-1)**n) / (2*n+1)
    pi_estimate = pi_estimate+fraction
    
    if (n+1)%100 == 0:
        pi_v = 4*pi_estimate
        relative_error = abs((pi_v-np.pi)/np.pi)
        print("Number of terms: ", n+1, "Pi estimate: ", pi_estimate, "Relative error: ", relative_error)
        
        #The estimated value of pi gets closer to the real value every 100 terms
        #The relative error decreases slowly
```
