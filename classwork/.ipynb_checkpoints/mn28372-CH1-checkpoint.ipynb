```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(1, 6)
y = np.array([2, 4, 6, 8, 10])
y2 = np.array([1, 3, 5, 7, 9])
rnumbers = np.random.randn(100)

plt.figure()
plt.plot(x, y, marker='o', color='blue')
plt.title('Line Graph')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()

plt.figure()
plt.scatter(x, y2, color='red')
plt.title('Scatter Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()

plt.figure()
plt.bar(x, y, color='green')
plt.title('Bar Graph')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()

plt.figure()
plt.hist(rnumbers, bins=10, color='purple')
plt.title('Histogram')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```
